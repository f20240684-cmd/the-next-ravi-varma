# 04 -- Generation

Generate Ravi Varma-inspired artwork using the trained LoRA, with optional ControlNet pose/edge conditioning.

```python
import sys, os
PROJECT_ROOT = os.path.abspath('..')
sys.path.insert(0, os.path.join(PROJECT_ROOT, 'src'))
os.chdir(PROJECT_ROOT)

from ravi_varma.config import GenerationConfig
from ravi_varma.generation.pipeline import RaviVarmaGenerator
from ravi_varma.generation.prompt_engine import PromptExpansionEngine
```

## Text-only generation

```python
config = GenerationConfig.load('configs/generation.yaml')
engine = PromptExpansionEngine(trigger_token=config.style.trigger_token, negative_prompt=config.negative_prompt)
expanded = engine.expand('Yudhishthira weeping over the fallen body of Karna on the battlefield of Kurukshetra.')
print(expanded.final_prompt())
```

```python
generator = RaviVarmaGenerator(config)
result = generator.generate(prompt=expanded.final_prompt(), negative_prompt=expanded.negative_prompt, seed=42)
result.image
```

## With a pose reference (ControlNet)

```python
from PIL import Image
pose_ref = Image.open('path/to/your_pose_reference.jpg')
result = generator.generate(
    prompt=expanded.final_prompt(),
    control_image=pose_ref,
    control_type='openpose',
    controlnet_strength=0.8,
    seed=42,
)
result.image
```

`result.metadata` contains every generation parameter plus whether a real LoRA checkpoint was loaded -- check `result.metadata['lora_loaded']`.
